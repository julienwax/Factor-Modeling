# Data processing of the 10 Years SPX daily returns:

```python
import pandas as pd
import warnings 
warnings.filterwarnings("ignore")
data = pd.read_csv("Data/SPX_daily_returns_10Y.csv")
```

```python
data.head(2)
```

```python
data.columns = [s.split()[0] for s in data.columns]
data = data.rename(columns = {"Unnamed:" : "Date"})
data = data.iloc[1:,:]
data["Date"] = pd.to_datetime(data["Date"], format = "%m/%d/%Y")
data.set_index("Date",inplace = True,drop=True)
data.head(2)
```

```python
data.tail()
```

```python
import pandas_market_calendars as mcal

calendars = mcal.get_calendar("NYSE")
business_days = calendars.schedule(start_date = '2016-03-09', end_date = '2026-03-09')
df = data[data.index.isin(business_days.index)]
```

# Analyzing sources of missingness: (mainly just dropping at first)

```python
# print(df.isna().mean().to_string())
# print(df.isna().sum()[df.isna().sum() > 0])
df = df[df.isna().sum()[df.isna().sum() == 0].index]
```

```python
# print(df.isna().mean().to_string())
print(df.shape)
# df.columns = df.columns.astype(str)
df.to_hdf("Data/daily_returns_SPX.h5", key = "returns")
```

- Now we have 2514 daily returns over the last 10 years of 418 US equities.
